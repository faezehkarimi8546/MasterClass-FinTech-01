```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('/content/drive/MyDrive/Billionaire.csv')
data.head()
```

```python
data.info()
data['Industry'].isna().sum()
```

```python
industry_counts = data['Industry'].value_counts()
industry_counts
```

```python
top_10 = industry_counts.head(10)
other_count = industry_counts.iloc[10:].sum()

final_counts = pd.concat([
    top_10,
    pd.Series({'Other': other_count})
])

print(final_counts)
```

```python
plt.figure(figsize=(10, 8))

colors = plt.get_cmap('tab20').colors[:11]
plt.pie(
    final_counts.values,
    autopct='%1.1f%%',
    startangle=90,
    colors=colors
)

plt.legend(
    final_counts.index,
    title='Industries',
    loc='center left',
    bbox_to_anchor=(1, 0.5)
)

plt.title('Distribution of Billionaires by Industry')

plt.show(

)
```

The dataset contains 2,755 billionaires across 18 industries. Finance & Investments has the highest number of billionaires, with 371 individuals.

To simplify the visualization, the 10 most common industries were displayed separately, while the remaining 8 industries were combined into a single category called "Other".

The Other category contains 377 billionaires, representing approximately 13.7% of the total.
